<h3>3. Data Cleaning </h3>

```python
#Removing the S N column from the DataFrame.
df = df.drop('S N', axis = 1)
df.head()
```

```python
#Filling missing values in Inches column using mode.
df_inches = df['Inches'].mode()[0]
```

```python
df['Inches'] = df['Inches'].fillna(df_inches)
```

```python
df['Inches'].isnull().sum()
```

```python
#Filling missing values in Ram column using mode.
df_Ram= df['Ram'].mode()[0]
```

```python
df['Ram'] = df['Ram'].fillna(df_Ram)
```

```python
df['Ram'].isnull().sum()
```

```python
#Verifying missing values handling using df.info()
df.info()
```

```python
#Convert Inches column to numeric type.
df['Inches'] = pd.to_numeric(df['Inches'], errors = 'coerce')
```

```python
df['Inches'].dtypes
```

```python
#Converting Ram column to numeric by removing text (GB).
df['Ram']= df['Ram'].str.replace('GB' ,'', regex = False).astype(int)
```

```python
df['Ram'].dtypes
```

```python
#Converting Weight column to numeric by removing text (kg)
df['Weight'] = df['Weight'].str.replace('kg' ,' ', regex=False).str.strip()
df['Weight'] = pd.to_numeric(df['Weight'], errors = 'coerce')
```

```python
df['Weight'].dtypes
```

```python
df.head(2)
```

<h3>4.Filtering & Selecting Data </h3>

```python
#Displaying all laptops from Dell company
df[df['Company'] == 'Dell']
```

```python
#Display laptops with Price greater than 1,00,000.
df[df['Price'] > 100000]
```

```python
#Display laptops running Windows OS.
df[df['OpSys'].str.contains('Windows', case = False, na = False)]
```

```python
#Displaying laptops with RAM greater than 8GB.
df[df['Ram'] > 8]
```

```python
#Counting the number of laptops per company.
df['Company'].value_counts()
```

<h3> 5. Creating New Columns</h3>

```python
#Creating a new column ‘Price in $’ by converting Price column into $ Dollars. 
df['Price in $'] = (df['Price'] / 83).round(2)
```

```python
df.head()
```

```python
#Creating Weight_Category column based on laptop weight.
df['Weight_Category'] = 'Heavy'
df.loc[df['Weight'] < 2, 'Weight_Category'] = 'Light'
```

```python
df.head(2)
```

<h3>6. Creating New DataFrames</h3>

```python
#Creating a DataFrame that contains Gaming laptops only.
gaming_lp = df[df['TypeName'].str.contains('Gaming', case=False, na=False)]
```
